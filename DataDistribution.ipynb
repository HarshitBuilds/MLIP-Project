```python
#python notebook used for ploting atomization energy (per atom) and other properties of different Mo-Cl clusters, from data in the extxyz file.
import matplotlib.pyplot as plt
import numpy as np
import re # Import the regular expression module

file_path = '/content/3mo6cl_1000_frames.extxyz'

energies = []
try:
    with open(file_path, 'r') as f:
        for line in f:
            # Check if the line contains "energy="
            if "energy=" in line:
                # Use regular expression to find the energy value
                match = re.search(r'energy=([-+]?\d*\.\d+)', line)
                if match:
                    energy = float(match.group(1))
                    energies.append(energy)

    if not energies:
        print("No energy values found in the file.")
    else:
        # Convert energies to a numpy array for easier handling
        energies_np = np.array(energies)
        energy_cl = -0.2503
        energy_mo = -4.587
        energies_np = energies_np - 6*energy_cl - 3*energy_mo #Atomization energy of the cluster
        energies_np = energies_np/9 #Atomization energy per atom
        # Plot a histogram of the atomization energy per atom.

        plt.figure(figsize=(10, 6))
        plt.hist(energies_np, bins=50, edgecolor='black')
        plt.xlabel('Energy (eV) per atom')
        plt.ylabel('Frequency')
        plt.title('Histogram of Energy Values')
        plt.grid(axis='y', alpha=0.75)
        plt.show()

except FileNotFoundError:
    print(f"Error: File not found at {file_path}")
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
# For 3Mo3Cl cluster
import matplotlib.pyplot as plt
import numpy as np
import re # Import the regular expression module

file_path_1 = '/content/3mo3cl.extxyz'

energies_1 = []
try:
    with open(file_path_1, 'r') as f:
        for line in f:
            # Use regular expression to find the energy value
            match = re.search(r'energy=([-+]?\d*\.\d+)', line)
            if match:
                energy = float(match.group(1))
                energies_1.append(energy)

    if not energies_1:
        print(f"No energy values found in the file: {file_path_1}")
    else:
        # Convert energies to a numpy array for easier handling
        energies_np_1 = np.array(energies_1)
        frames_1 = np.arange(len(energies_np_1))

        # Plot energy vs frame
        plt.figure(figsize=(12, 6))
        plt.plot(frames_1, energies_np_1, marker='o', linestyle='-', markersize=2)
        plt.xlabel('Frame Number')
        plt.ylabel('Energy (eV)')
        plt.title('Energy vs Frame Number (3mo3cl.extxyz)')
        plt.grid(True)
        plt.show()

except FileNotFoundError:
    print(f"Error: File not found at {file_path_1}")
except Exception as e:
    print(f"An error occurred while processing {file_path_1}: {e}")

# plotting magnetic moment
import matplotlib.pyplot as plt
import numpy as np
import re # Import the regular expression module

file_path_1 = '/content/3mo3cl.extxyz'

magmom_1 = [] # Changed variable name to magmom_1
try:
    with open(file_path_1, 'r') as f:
        for line in f:
            # Use regular expression to find the magnetic moment value
            match = re.search(r'magmom=([-+]?\d*\.\d+)', line) # Changed regex to search for magmom
            if match:
                magmom = float(match.group(1))
                magmom_1.append(magmom) # Appended to magmom_1

    if not magmom_1: # Checked magmom_1
        print(f"No magnetic moment values found in the file: {file_path_1}")
    else:
        # Convert magnetic moments to a numpy array for easier handling
        magmom_np_1 = np.array(magmom_1) # Changed variable name
        frames_1 = np.arange(len(magmom_np_1)) # Used magmom_np_1 to get length

        # Plot magnetic moment vs frame
        plt.figure(figsize=(12, 6))
        plt.plot(frames_1, magmom_np_1, marker='o', linestyle='-', markersize=2) # Plotted magmom_np_1
        plt.xlabel('Frame Number')
        plt.ylabel('Magnetic Moment') # Changed y-label
        plt.title('Magnetic Moment vs Frame Number (3mo3cl.extxyz)') # Changed title
        plt.grid(True)
        plt.show()

except FileNotFoundError:
    print(f"Error: File not found at {file_path_1}")
except Exception as e:
    print(f"An error occurred while processing {file_path_1}: {e}")
```
